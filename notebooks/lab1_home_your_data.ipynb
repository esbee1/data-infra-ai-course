```python
FILE_NAME = "your_file.csv"   # <- change this to your file's name
```

```python
from pathlib import Path
import duckdb

ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
DATA = ROOT / "data" / FILE_NAME

if not DATA.exists():
    print(f"Can't find data/{FILE_NAME}. Files in data/ right now:")
    for f in sorted((ROOT / "data").iterdir()):
        print("  ", f.name)
else:
    con = duckdb.connect(str(ROOT / "lab1_home.duckdb"))
    print(f"Found data/{FILE_NAME} ({DATA.stat().st_size / 1_000_000:.1f} MB). Connected.")
```

## 3. First look

## 4. Load it and check the types

Load the file into a table called `my_data`, then `DESCRIBE` it. Compare each column's type with what it *should* be. A number column stored as `VARCHAR`, or a date column stored as text, is the same kind of problem you fixed in class.

```python
con.sql(f"CREATE OR REPLACE TABLE my_data AS SELECT * FROM read_csv('{DATA}')")
print(con.sql("SELECT count(*) FROM my_data").fetchone()[0], "rows loaded")
con.sql("DESCRIBE my_data")
```

`SUMMARIZE` gives a profile of every column at once: min, max, how many distinct values, and `null_percentage` (how much is missing).

```python
con.sql("SUMMARIZE my_data")
```

## 5. Fix what loaded wrong (if anything)

Otherwise, open the CSV (double-click it in the `data` folder) and find **how missing values are written** in the columns that came in as text. Then edit the `nullstr` list below to match and run the cell. Common markers: `NA`, `N/A`, `-`, `null`, `(blank)`.
